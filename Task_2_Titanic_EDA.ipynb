```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"], bins=30, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Fare"], bins=30, kde=True)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.histplot(df["Pclass"], discrete=True)

plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Age"])

plt.title("Age Boxplot")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Fare"])

plt.title("Fare Boxplot")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(x="Survived", data=df)

plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(x="Sex", hue="Survived", data=df)

plt.title("Survival by Gender")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(x="Pclass", hue="Survived", data=df)

plt.title("Survival by Passenger Class")

plt.show()
```

```python
sns.pairplot(
    df[["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]],
    hue="Survived"
)

plt.show()
```

```python
plt.figure(figsize=(10,7))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

## EDA Findings

1. The Titanic dataset contains passenger information such as age, passenger class, gender, fare and survival status.

2. The age distribution shows that passengers were mostly concentrated in the younger and middle-age groups.

3. The fare distribution contains some high-value observations that appear as potential outliers in the boxplot.

4. The survival count shows that the number of passengers who did not survive was higher than the number who survived.

5. Gender appears to have a noticeable relationship with survival.

6. Passenger class also appears to have a relationship with survival, with different survival patterns across the three classes.

7. The correlation matrix helps identify relationships between numerical features.

8. EDA helps us understand patterns, distributions, relationships and possible anomalies before applying machine learning models.
